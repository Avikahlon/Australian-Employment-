## load the labour force dataset and extract the unemployment rate of different states. 


```python
import pandas as pd
```

```python
file_path = r"./datatset/6202012-Labour_force.xlsx"
df = pd.read_excel(file_path, sheet_name=2)
```

```python
df.head()
```

```python
# Filter columns that contain 'Employment' in their names and drop them
labour_force_df = df
columns_to_drop = [col for col in labour_force_df.columns if 'Employment' in col]
labour_force_df_cleaned = labour_force_df.drop(columns=columns_to_drop)

# Check the first few rows of the cleaned dataframe
labour_force_df_cleaned.head()
```

```python
# Get the columns that have "Trend" or "Seasonally Adjusted" in row 1 (which is index 1 in pandas)
columns_to_drop = labour_force_df_cleaned.columns[(labour_force_df_cleaned.iloc[1] == "Trend") | (labour_force_df_cleaned.iloc[1] == "Seasonally Adjusted")]

# Drop the identified columns
labour_force_df_cleaned = labour_force_df_cleaned.drop(columns=columns_to_drop)

# Check the first few rows of the cleaned dataframe
labour_force_df_cleaned.head()
```

```python
output_path = r"./datatset/unemployment_of_all_states.csv"
labour_force_df_cleaned.to_csv(output_path, index=False)
```

```python
import pandas as pd

# Load the Excel file
file_path = "./datatset/6291005-employmnet_rate_of_different_industry.xlsx"
df = pd.read_excel(file_path, sheet_name=2)

# Filter columns that contain 'Western Australia' and exclude those with 'part-time' or 'full-time'
filtered_df = df.loc[:, df.columns.str.contains("Western Australia", case=False) & 
                     ~df.columns.str.contains("part-time|full-time", case=False)]

# Add the first column of the original df to filtered_df
filtered_df.insert(0, df.columns[0], df.iloc[:, 0])

# Save the result to a CSV file
filtered_df.to_csv("wa_employment_data.csv", index=False)

print("Filtered data saved as 'filtered_employment_data.csv'")
```

```python
import pandas as pd

# Load the filtered CSV and the original Excel file (sheet 1 for additional columns)
filtered_path = "./datatset/employment_data.csv"
original_path = "./datatset/6291005-employmnet_rate_of_different_industry.xlsx"
filtered_df = pd.read_csv(filtered_path)
df_sheet1 = pd.read_excel(original_path, sheet_name=1)

# Convert date columns in both DataFrames to a common date format (ignoring the time)
filtered_df['date'] = pd.to_datetime(filtered_df['Unnamed: 0'], errors='coerce').dt.date
df_sheet1['date'] = pd.to_datetime(df_sheet1['Unnamed: 0'], errors='coerce').dt.date

# Merge the DataFrames on the 'date' column
merged_df = pd.merge(filtered_df, df_sheet1.iloc[:, :59], on="Unnamed: 0", how="left")

# Remove columns that contain 'part-time' or 'full-time'
cleaned_df = merged_df.loc[:, ~merged_df.columns.str.contains("part-time|full-time", case=False)]

# Save the cleaned DataFrame to a new CSV
cleaned_df.to_csv("./datatset/employment_data.csv", index=False)

print("Cleaned data saved as 'cleaned_employment_data.csv'")
```
